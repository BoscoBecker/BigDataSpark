```python
# abre sessão no spark
import pyspark
conf = pyspark.SparkConf()
conf.setMaster('local[*]')
sc = pyspark.SparkContext.getOrCreate(conf)
```

```python
#Efetua download do arquivo covid
!gdown 1SRY5fDLNCc3DPzIUr7-9af0daWd1LkNm
```

```python
#carrega arquivo da COVID em um rdd
rdd = sc.textFile('file:////content/covid_sample.csv')
```

```python
def GetFields(line):
  fields = line.split(';')
  id  = fields[0]
  dataNotificacao = fields[1]
  dataInicioSintomas = fields[2]
  dataNascimento = fields[3]
  sintomas = fields[4]
  profissionalSaude = fields[5]
  cbo = fields[6]
  condicoes = fields[7]
  estadoTeste = fields[8]
  dataTeste = fields[9]
  tipoTeste = fields[10]
  resultadoTeste = fields[11]
  paisOrigem = fields[12]
  sexo = fields[13]
  bairro = fields[14]
  estado = fields[15]
  estadoIBGE = fields[16]
  municipio = fields[17]
  municipioIBGE = fields[18]
  cep = fields[19]
  origem = fields[20]
  cnes = fields[21]
  estadoNotificacao = fields[22]
  estadoNotificacaoIBGE = fields[23]
  municipioNotificacao = fields[24]
  municipioNotificacaoIBGE = fields[25]
  numeroNotificacao = fields[26]
  excluido = fields[27]
  validado = fields[28]
  idade = fields[29]
  dataEncerramento = fields[30]
  evolucaoCaso = fields[31]
  classificacaoFinal = fields[32]
  return id, dataNotificacao, dataInicioSintomas, dataNascimento, sintomas, profissionalSaude, cbo, condicoes, estadoTeste, dataTeste, tipoTeste, resultadoTeste, paisOrigem, sexo, bairro, estado, estadoIBGE, municipio, municipioIBGE, cep, origem, cnes, estadoNotificacao, estadoNotificacaoIBGE, municipioNotificacao, municipioNotificacaoIBGE, numeroNotificacao, excluido, validado, idade, dataEncerramento, evolucaoCaso,
  classificacaoFinal

def isGenderMale(fields):
   return 1 if fields[13] == 'Feminino' else 0
```

```python
fields = rdd.map(GetFields).cache()
```

```python
# 1 - Quantidade de pacientes positivos para corona virus (resultadoTeste)
fields.filter(lambda x: x[11] == 'Positivo').count()
```

```python
# 2 - Quantidade de pacientes do sexo feminino positivos para corona virus (resultadoTeste)
fields.filter(isGenderMale).filter(lambda x: x[11] == 'Positivo').count()
```

```python
 # 3 - Quantidade de pacientes de acordo com o sexo e o resultado do teste (resultadoTeste), por positivos e negativos
resultado = fields.filter(lambda x: x[11] == 'Positivo').map(lambda x: x[13]).countByValue().items()

resultado = (
    fields
    .map(lambda x: (x[13], x[11]))
    .countByValue()
    .items()
)

for (sexo, resultadoTeste), quantidade in resultado:
  print(f"{sexo} - {resultadoTeste}: {quantidade}")
```

```python
# - 4 Sintomas mais comuns para casos positivos para corona virus
fields.filter(lambda x: x[11] == 'Positivo').map(lambda x: x[4]).countByValue().items()
```

```python
# - 5 Sintomas mais comuns para casos não positivos para corona virus
fields.filter(lambda x: x[11] != 'Positivo').map(lambda x: x[4]).countByValue().items()
```

```python
# 6 - Quantidade de casos positivos no Paraná
fields.filter(lambda x: x[11] == 'Positivo').filter(lambda x: x[15] == 'PARANÁ').count()
```

```python
 # 7 - Municipio do Paraná com a maior quantidade de óbitos (evolucaoCaso)
fields.filter(lambda x: x[31] == 'Óbito').filter(lambda x: x[15] == 'PARANÁ').map(lambda x: x[17]).countByValue().items()
```

```python
# 8 - Quantidade de municipios no Paraná com casos positivos
fields.filter(lambda x: x[11] == 'Positivo').filter(lambda x: x[15] == 'PARANÁ').map(lambda x: x[17]).distinct().count()
```

```python
# 9 - Quantidade de municipios no Paraná sem casos positivos
fields.filter(lambda x: x[11] != 'Positivo').filter(lambda x: x[15] == 'PARANÁ').map(lambda x: x[17]).distinct().count()
```

```python
# 10 - Estado com a maior taxa de falecimento de mulheres
fields.filter(isGenderMale).filter(lambda x: x[31] == 'Óbito').map(lambda x: x[15]).countByValue().items()
```

```python
# 11 - Menor idade de mulher positiva para covid
fields.filter(isGenderMale).filter(lambda x: x[11] == 'Positivo').map(lambda x: x[29]).min()
```

```python
# 12 - Maior idade de mulher positiva para covid
fields.filter(isGenderMale).filter(lambda x: x[11] == 'Positivo').map(lambda x: x[29]).max()
```

```python
# 13 - Quantidade de casos positivos, em granularidade diária, ao longo do tempo
fields.filter(lambda x: x[11] == 'Positivo').map(lambda x: x[9]).countByValue().items()
```

```python
# 14 - Quantidade de casos positivos, em granularidade semanal, ao longo do tempo
fields.filter(lambda x: x[11] == 'Positivo').map(lambda x: x[9]).map(lambda x: x[:10]).countByValue().items()
```

```python
# 15 -  Quantidade de pacientes que faleceram (evolucaoCaso) separados pela idade
fields.filter(lambda x: x[31] == 'Óbito').map(lambda x: x[29]).countByValue().items()
```

```python
# 16 - Taxa de óbito (evolucaoCaso) por idade
fields.filter(lambda x: x[31] == 'Óbito').map(lambda x: x[29]).countByValue().items()
```

```python
# 17 - Idade média das mulheres positivas para covid
media = fields.filter(isGenderMale).filter(lambda x: x[11] == 'Positivo').map(lambda x: float(x[29])).mean()
print(media)
```

```python
# 18 -Município do Paraná com a maior quantidade de mulheres positivos para covid
fields.filter(isGenderMale).filter(lambda x: x[11] == 'Positivo').filter(lambda x: x[15] == 'PARANÁ').map(lambda x: x[17]).countByValue().items()
```

```python
# 19 - Dia da semana com a maior quantidade de testes realizados
resultado = fields.map(lambda x: x[9][:3]).countByValue()
print(list(resultado.items()))
```

```python
# 20 - Dia da semana com maior quantidade de pacientes positivos para corona virus
resultado  = fields.filter(lambda x: x[11] == 'Positivo').map(lambda x: x[9][:3]).countByValue()
print(list(resultado.items()))
```

```python
# 21 -  Municipio, com mais de 500 testes, com a maior taxa de exames que não foram positivos para COVID (casos que foram testados porém não tiveram o resultadoTeste como Positivo)
fields.filter(lambda x: x[11] != 'Positivo').map(lambda x: x[17]).countByValue().items()
```
